## invoke和format返回的数据类型不同
> 我的邻居是tom, 兴趣是篮球 <class 'str'>
> text='我的邻居是jack, 兴趣是象棋' <class 'langchain_core.prompt_values.StringPromptValue'>

```python
from langchain_core.prompts import PromptTemplate, FewShotPromptTemplate, ChatPromptTemplate

template = PromptTemplate.from_template("我的邻居是{name}, 兴趣是{hobby}")

res = template.format(name="tom", hobby="篮球")

print(res, type(res))

res2 = template.invoke({"name": "jack", "hobby": "象棋"})

print(res2, type(res2))
```
