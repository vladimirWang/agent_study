```python
from langchain_community.document_loaders import JSONLoader

loader = JSONLoader(
    file_path="./stu.json",
    jq_schema=".name"
)

docs = loader.load()

for doc in docs:
    print(doc.page_content)
    print(type(doc), doc)

loader2 = JSONLoader(
    file_path="./stu.json",
    jq_schema=".info",
    text_content=False
)

docs = loader2.load()

for doc in docs:
    print(doc.page_content)
    print(type(doc), doc)
```
