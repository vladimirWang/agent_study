```python
import numpy as np

def get_dot(vec_a, vec_b):
    if len(vec_a) != len(vec_b):
        raise ValueError("长度必须相同")
    
    result = 0;
    for a, b in zip(vec_a, vec_b):
        result+= a*b;
    return result

dot1 = get_dot([0.5,0.5], [0.7,0.7])
print(dot1)
 
def get_norm(vec):
    result = np.sqrt(vec[0]*vec[0] + vec[1]*vec[1])
    return result

def cosine_similarity(vec_a, vec_b):
    return get_dot(vec_a, vec_b) / (get_norm(vec_a) *get_norm(vec_b))
similarity = cosine_similarity([0.5,0.5], [0.7,0.7])
print(similarity)
```
